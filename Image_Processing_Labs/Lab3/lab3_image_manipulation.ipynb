### 1. Scale (increase size)
Double the image dimensions using `Image.resize()` with high-quality resampling `(LANCZOS)`.

```python
# Get the size of the image
width, height = img.size
# scaling factors
scale_factor = 2
```

```python
# ── 1. Increase size (scale ×2) ────────────────────
new_width = width * scale_factor
new_height = height * scale_factor
# Resize the image using PIL's built-in method
new_size = (new_width, new_height)
scaled_image = img.resize(new_size, Image.Resampling.LANCZOS)

```

```python

# Save the scaled image and print the sizes (The new image name should be "task1_1_scaled.jpg")
scaled_image.convert('RGB').save("task1_1_scaled.jpg")
print("Original size:", img.size)
print("Scaled size:", scaled_image.size)
```

### 3. Shear

```python
# -- c. Get the image dimensions ────────────────────
width, height = img.size
```

```python
# -- d. define the shear matrix ──────────────────────────
# Choose X-axis or Y-axis shear. The shear factor controls how much the image slants — start with 0.5 then experiment.
shx = 0.5  
shy = 0.0
```

```python
# -- e. Apply the shear transformation to the image ──────────────────────────
# PIL's transform() takes the inverse affine matrix:
# [ 1    shx   tx ]
# [ shy  1     ty ]
# e. Apply the shear transformation to the image
new_width = int(width + abs(shx) * height)


matrix = (1, -shx, 0, 0, 1, 0)

sheared_image = img.convert('RGB').transform(
    (new_width, height),
    Image.AFFINE,
    matrix,
    resample=Image.BICUBIC
)
```

```python
# -- f. Save the sheared image(The new image name should be "task1_3_sheared.jpg") 
sheared_image.convert('RGB').save("task1_3_sheared.jpg")
print("Original size:", img.size)
print("Sheared size:", sheared_image.size)
```

### Experiment and compare
Try these:

— Change shear factor from 0.5 to 0.1, 0.3, 0.8 and compare results

— Switch from X-axis to Y-axis shear matrix

— Update the canvas multiplier to match your new factor

— Try combining X and Y shear in one matrix

```python
# ── Experiment and Compare 

# 1. Change shear factor to 0.1, 0.3, and 0.8 on X-axis
for factor in [0.1, 0.3, 0.8]:
    new_w = int(width + abs(factor) * height)  
    matrix_x = (1, -factor, 0, 0, 1, 0)
    sh_img = img.convert("RGB").transform(
        (new_w, height), Image.AFFINE, matrix_x, resample=Image.BICUBIC
    )
    sh_img.save(f"task1_3_sheared_x_{factor}.jpg")

# 2. Switch to Y-axis shear matrix 
shy = 0.5
new_h = int(height + abs(shy) * width)  
matrix_y = (1, 0, 0, -shy, 1, 0)
sheared_y = img.convert("RGB").transform(
    (width, new_h), Image.AFFINE, matrix_y, resample=Image.BICUBIC
)
sheared_y.save("task1_3_sheared_y_0.5.jpg")

# 3. Combine X and Y shear in one matrix
shx_comb, shy_comb = 0.3, 0.3
new_w_comb = int(width + abs(shx_comb) * height)
new_h_comb = int(height + abs(shy_comb) * width)
matrix_xy = (1, -shx_comb, 0, -shy_comb, 1, 0)

sheared_comb = img.convert("RGB").transform(
    (new_w_comb, new_h_comb),
    Image.AFFINE,
    matrix_xy,
    resample=Image.BICUBIC,
)
sheared_comb.save("task1_3_sheared_combined.jpg")
```

# Intensity Transformations
Negative · Log · Power Law (Gamma)

```python

# ── 1. Negative ──────────────────────────────────── 
# Method 1: NumPy array manipulation

import numpy as np

img_arr = np.array(img.convert('RGB'))
neg_arr = 255 - img_arr


negative_img1 = Image.fromarray(neg_arr)
negative_img1.save("task2_1_negative_numpy.jpg")
```

```python
# Method 2: PIL's ImageOps
from PIL import ImageOps

negative_img2 = ImageOps.invert(img.convert('RGB'))
negative_img2.save("task2_1_negative_pil.jpg")
```

```python
# ── 2. Log transformation ──────────────────────────

# s = c · log(1 + r) --> We need to find c.
# We want the maximum output value (s) to be 255 when the maximum input value (r) is 255:
# 255 = c · log(1 + 255)
# c = 255 / log(1 + 255)

img_arr = np.array(img.convert('RGB'), dtype=np.float64)
c = 255 / np.log(1 + 255)


log_transformed_arr = c * np.log(1 + img_arr)


log_image = Image.fromarray(np.uint8(log_transformed_arr))
log_image.save("task2_2_log.jpg")


# Apply the log transformation to each pixel
```

```python

# ── 3. Power-law / Gamma correction ───────────────

gamma = 0.5  
c = 1


gamma_arr = c * np.power(img_arr / 255.0, gamma) * 255.0


gamma_image = Image.fromarray(np.uint8(gamma_arr))
gamma_image.save("task2_3_gamma.jpg")
```
